Loading Dataset

```python
# Imporing necessary libraries
import pandas as pd
import numpy as np

# Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Defining the file path
file_path = '/content/drive/MyDrive/Customer Insights and Data Analytics Early Remote Internship/Freshdesk Customer Support Tickets Dataset Filled.csv'

# Reading the CSV file
df = pd.read_csv(file_path)

# Displaying first few rows to verify
df.head()
```

Data Extraction & Initial Assessment

```python
# Checking number of rows and columns
rows, columns = df.shape
print(f"Number of rows: {rows}")
print(f"Number of columns: {columns}")
```

```python
# Checking for duplicate rows
duplicate_rows = df[df.duplicated()]
print(f"Number of duplicate rows: {duplicate_rows.shape[0]}")
```

```python
# Checking for null values in each column
null_counts = df.isnull().sum()
print("Null values in each column:")
print(null_counts)
```

```python
# Checking data types of all columns
df.dtypes
```

```python
# Converting 'Created Date' and 'Resolved Date' to datetime
df['Created Date'] = pd.to_datetime(df['Created Date'], format='%Y-%m-%d %H:%M')
df['Resolved Date'] = pd.to_datetime(df['Resolved Date'], format='%Y-%m-%d %H:%M')

# Verify the conversion
df.dtypes
```

```python
# Check unique values in key categorical columns
categorical_columns = ['Category', 'Priority', 'Status', 'Tags', 'Sentiment']

for col in categorical_columns:
    print(f"Unique values in '{col}':")
    print(df[col].unique())
    print("-" * 50)
```

```python
# Stripping whitespace and ensuring consistent capitalization in 'Sentiment'
df['Sentiment'] = df['Sentiment'].str.strip().str.capitalize()

# Checking again
print(df['Sentiment'].unique())
```

```python
# Exporting csv file
output_path = "/content/drive/MyDrive/Customer Insights and Data Analytics Early Remote Internship/Cleaned_Freshdesk_Customer_Support_Tickets_Dataset_Filled.csv"

# Saving as CSV
df.to_csv(output_path, index=False)

print(f"Cleaned dataset successfully saved to: {output_path}")
```
